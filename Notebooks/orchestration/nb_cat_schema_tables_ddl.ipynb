### ** Catalog · Schema · Volume — IntelliBI Sales Lakehouse **

Single source of truth for all object provisioning.  Runs in **dev**, **qa**, **prd**.

* Existing entities (Customer, Sales, Discount, Invoice) — unchanged.
* **New entities (Region, Product, Store + Dimensions + Facts) point to AWS S3** at:
  `s3://saleslake-748005667258-eu-north-1-an/saleslake/<env>/delta_tables/<layer>/<entity>/`
* All duplicates from earlier revisions removed.

```python
%sql
-------------- Create Catalogs for all environments --------------  
CREATE CATALOG IF NOT EXISTS saleslake_dev;
CREATE CATALOG IF NOT EXISTS saleslake_qa;
CREATE CATALOG IF NOT EXISTS saleslake_prd;

-------------- Create Schemas (bronze / silver / gold per env) --------------
CREATE SCHEMA IF NOT EXISTS saleslake_dev.bronze_schema_dev;
CREATE SCHEMA IF NOT EXISTS saleslake_dev.silver_schema_dev;
CREATE SCHEMA IF NOT EXISTS saleslake_dev.gold_schema_dev;
CREATE SCHEMA IF NOT EXISTS saleslake_qa.bronze_schema_qa;
CREATE SCHEMA IF NOT EXISTS saleslake_qa.silver_schema_qa;
CREATE SCHEMA IF NOT EXISTS saleslake_qa.gold_schema_qa;
CREATE SCHEMA IF NOT EXISTS saleslake_prd.bronze_schema_prd;
CREATE SCHEMA IF NOT EXISTS saleslake_prd.silver_schema_prd;
CREATE SCHEMA IF NOT EXISTS saleslake_prd.gold_schema_prd;

---------- Create Volumes (source-file loc landing zone) -----------
CREATE VOLUME IF NOT EXISTS saleslake_dev.bronze_schema_dev.vol_saleslake_src_files_dev;
CREATE VOLUME IF NOT EXISTS saleslake_qa.bronze_schema_qa.vol_saleslake_src_files_qa;
CREATE VOLUME IF NOT EXISTS saleslake_prd.bronze_schema_prd.vol_saleslake_src_files_prd;
```
